```python
import pandas as pd
import numpy as np
import folium
from folium.plugins import MarkerCluster

# Gerando dados sintéticos de imóveis em Nova Iguaçu e Queimados
np.random.seed(42)
n_imoveis = 45

dados_imoveis = {
    'id_imovel': range(1, n_imoveis + 1),
    'cidade': np.where(np.random.rand(n_imoveis) > 0.4, 'Nova Iguaçu', 'Queimados'),
    'valor_venda': np.random.uniform(150000, 850000, n_imoveis).round(2),
    'tipo': np.random.choice(['Casa', 'Apartamento', 'Terreno'], n_imoveis)
}

df_mapa = pd.DataFrame(dados_imoveis)

# Atribuindo coordenadas com base na cidade com dispersão aleatória
def gerar_lat(cidade):
    if cidade == 'Nova Iguaçu':
        return -22.756 + np.random.uniform(-0.03, 0.03)
    return -22.716 + np.random.uniform(-0.02, 0.02)

def gerar_lon(cidade):
    if cidade == 'Nova Iguaçu':
        return -43.460 + np.random.uniform(-0.03, 0.03)
    return -43.555 + np.random.uniform(-0.02, 0.02)

df_mapa['latitude'] = df_mapa['cidade'].apply(gerar_lat)
df_mapa['longitude'] = df_mapa['cidade'].apply(gerar_lon)

# Exibindo as primeiras linhas para validação
df_mapa.head()
```

```python
# Calculando a coordenada média de todos os imóveis para centralizar o mapa
lat_media = df_mapa['latitude'].mean()
lon_media = df_mapa['longitude'].mean()

# Criando o mapa base
mapa_basico = folium.Map(location=[lat_media, lon_media], zoom_start=12)

# Iterando sobre as 5 primeiras linhas e adicionando marcadores simples
for index, row in df_mapa.head(5).iterrows():
    popup_texto = f"Tipo: {row['tipo']}<br>Valor: R$ {row['valor_venda']:,.2f}"

    folium.Marker(
        location=[row['latitude'], row['longitude']],
        popup=folium.Popup(popup_texto, max_width=300)
    ).add_to(mapa_basico)

# Exibindo o mapa no Colab
mapa_basico
```

```python
# Criando novo mapa base
mapa_circulos = folium.Map(location=[lat_media, lon_media], zoom_start=12)

# Adicionando marcadores circulares para todos os imóveis
for index, row in df_mapa.iterrows():
    # Definindo a cor de preenchimento conforme a regra de negócio
    cor = 'blue' if row['cidade'] == 'Nova Iguaçu' else 'orange'

    folium.CircleMarker(
        location=[row['latitude'], row['longitude']],
        radius=8,
        color=cor,
        fill=True,
        fill_color=cor,
        fill_opacity=0.7,
        tooltip="Clique para detalhes",
        popup=f"Cidade: {row['cidade']}<br>Tipo: {row['tipo']}<br>Valor: R$ {row['valor_venda']:,.2f}"
    ).add_to(mapa_circulos)

# Exibindo o mapa no Colab
mapa_circulos
```

```python
# Criando o terceiro mapa
mapa_cluster = folium.Map(location=[lat_media, lon_media], zoom_start=12)

# Instanciando o objeto MarkerCluster
marker_cluster = MarkerCluster().add_to(mapa_cluster)

# Adicionando os imóveis ao cluster com ícones customizados baseados no tipo
for index, row in df_mapa.iterrows():
    # Definindo a cor do ícone conforme o tipo de imóvel
    if row['tipo'] == 'Casa':
        cor_icone = 'green'
    elif row['tipo'] == 'Apartamento':
        cor_icone = 'blue'
    else:
        cor_icone = 'cadetblue' # Usado como equivalente a cinza/neutro suportado pelo folium

    popup_texto = f"<b>ID:</b> {row['id_imovel']}<br><b>Tipo:</b> {row['tipo']}<br><b>Cidade:</b> {row['cidade']}<br><b>Valor:</b> R$ {row['valor_venda']:,.2f}"

    folium.Marker(
        location=[row['latitude'], row['longitude']],
        popup=folium.Popup(popup_texto, max_width=300),
        icon=folium.Icon(color=cor_icone, icon='home', prefix='fa')
    ).add_to(marker_cluster)

# Salvando o mapa final em arquivo HTML
mapa_cluster.save('mapa_imoveis_baixada.html')

# Exibindo o mapa final no Colab
mapa_cluster
```
